```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv("/content/drive/MyDrive/used_cars.csv")

columns = ["brand", "model_year", "milage", "fuel_type", "transmission", "accident", "clean_title", "price"]

data = data[columns]
data
```

```python
from collections import Counter
data["clean_title"] = data["clean_title"].replace({np.nan: False, "Yes": True})
filtered_data = data.dropna()
cars_per_brand = Counter(data["brand"])
keep_brands = [k for k, v in cars_per_brand.items() if v >= 20]
filtered_data = filtered_data[filtered_data['brand'].isin(keep_brands)]

```

```python
filtered_data["milage"] = filtered_data["milage"].apply(lambda x: int(x[:-4].replace(",", "")))
filtered_data["price"] = filtered_data["price"].apply(lambda x: int(x[1:].replace(",", "")))

non_numeric_columns = filtered_data.select_dtypes(exclude=['number']).columns
filtered_data = pd.get_dummies(filtered_data, columns = non_numeric_columns, drop_first=True)
filtered_data
```

```python
filtered_data.describe()
```

```python
import numpy as np
import pandas as pd
import scipy.io as io
from sklearn import metrics
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Lasso, Ridge
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
```

```python
X = filtered_data.drop(columns=['price'])
y = filtered_data['price']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
def train_model(model):
  model.fit(X_train_scaled, y_train)
  print("=" * 30)
  print(model)
  y_pred_scaled = model.predict(X_train_scaled)
  rmse_scaled = mean_squared_error(y_train, y_pred_scaled) ** 0.5
  mae_scaled = mean_absolute_error(y_train, y_pred_scaled)
  r2_scaled = r2_score(y_train, y_pred_scaled)
  print("train:")
  print(f"\tRMSE: {rmse_scaled}")
  print(f"\tMAE: {mae_scaled}")
  print(f"\tR2: {r2_scaled}")
  y_pred_scaled = model.predict(X_test_scaled)
  rmse_scaled = mean_squared_error(y_test, y_pred_scaled) ** 0.5
  mae_scaled = mean_absolute_error(y_test, y_pred_scaled)
  r2_scaled = r2_score(y_test, y_pred_scaled)
  print()
  print("test:")
  print(f"\tRMSE: {rmse_scaled}")
  print(f"\tMAE: {mae_scaled}")
  print(f"\tR2: {r2_scaled}")
  print("=" * 30)
  print()
  return model
```

```python
train_model(LinearRegression())
train_model(Lasso(1))
train_model(Ridge(1))
```

```python
param_grid = {'alpha': [0] + [10 ** x for x in range(-5, 6)]}
print(param_grid)

print("BEST MODELS")
print()
lasso_gs = GridSearchCV(Lasso(random_state=0), param_grid, scoring='neg_mean_squared_error', cv=3)
lasso_gs.fit(X_train_scaled, y_train)
best_alpha = lasso_gs.best_estimator_.alpha
lasso_model = train_model(Lasso(best_alpha))

ridge_gs = GridSearchCV(Ridge(random_state=0), param_grid, scoring='neg_mean_squared_error', cv=3)
ridge_gs.fit(X_train_scaled, y_train)
best_alpha = ridge_gs.best_estimator_.alpha
ridge_model = train_model(Ridge(best_alpha))
```

Данные обработаны: категориальные признаки закодированы (One-Hot), числовые — отмасштабированы (StandardScaler).

Базовая модель (LinearRegression): показала R² = 0.41 на тесте.

Регуляризация: Lasso и Ridge с параметрами по умолчанию (alpha=1) не дали прироста качества.

Подбор гиперпараметров (GridSearchCV): выявил, что моделям нужна крайне сильная регуляризация (alpha=1000).

Финальный результат: лучшей стала Ridge (alpha=1000) с R² = 0.458. Прирост качества по сравнению с базовой моделью составил около +5 п.п.
